**Custom function for running pairwise meta-analysis** using `meta` and `netmeta`.

```r
pairwise_meta_analysis <- function(df, sm = "") {
    data_path <- paste0("./input/data/sheets_for_analysis/", df, ".csv")
    data <- read.csv(data_path)

    m.gen <- metagen(
        data = data,
        studlab = study,
        TE = TE,
        seTE = seTE,
        sm = paste0(sm),
        random = TRUE,
        common = FALSE,
        subgroup = subgroup
    )

    assign(paste0(df, "_pw"), m.gen, envir = .GlobalEnv)

    # Create data frame
    dat <- data.frame(
        "k" = round(m.gen$k.w, 4),
        sm = round(m.gen$TE.random.w, 4),
        "ci.lb" = round(m.gen$lower.random.w, 4),
        "ci.ub" = round(m.gen$upper.random.w, 4),
        "pval" = round(m.gen$pval.random.w, 4),
        "I^2" = round(m.gen$I2.w, 4)
    )

    write.csv(dat, paste0("./output/tables/pairwise/", df, ".csv"))
    assign(paste0(df, "_dat"), dat, envir = .GlobalEnv)
    print(dat)
}
```

```r
continuous = c("ikdc_subjective", "lysholm", "tegner", "instrumented_laxity")
lapply(continuous, sm = "SMD", pairwise_meta_analysis)
```

```r
dichotomous = c("pivot_shift_1", "pivot_shift_2", "lachman_1", "lachman_2", "graft_failure")
lapply(dichotomous, sm = "OR", pairwise_meta_analysis)
```

**Custom function for running network meta-analysis** using `netmeta`.

```r
network_meta_analysis <- function(df, sm, reference = "", backtransf = FALSE) {
    
    data_path <- paste0(root, "/input/data/sheets_for_analysis/", df, ".csv")
    data <- read.csv(data_path)
    
    m.netmeta <- netmeta(
        data = data,
        studlab = study,
        TE = TE,
        treat1 = treat1,
        treat2 = treat2,
        seTE = seTE,
        sm = paste0(sm),
        reference.group = paste0(reference),
        random = TRUE,
        common = FALSE,
        backtransf = TRUE,
        sep.trts = " vs. "
        )
    
    # results (netmeta, decomp.design, netsplit, netleague, netrank)
    sink(file = paste0("./output/results/netmeta/", df, ".txt"))
    print(m.netmeta)
    sink()

    sink(file = paste0("./output/results/decomp/", df, ".txt"))
    decomp <- decomp.design(m.netmeta)
    print(decomp)
    sink()

    sink(file = paste0("./output/results/netsplit/", df, ".txt"))
    netsplit <- netsplit(m.netmeta)
    print(netsplit)
    sink()

    sink(file = paste0("./output/results/netrank/", df, ".txt"))
    netrank <- netrank(m.netmeta,
                        method = "SUCRA",
                        nsim = 10000,
                        small.values = "bad"
                       )
    print(netrank)
    sink()

    order =  c("BPTB", "HT", "QT", "PLT", "AT", "TA")
    seq = order[order %in% m.netmeta$trts]
        
    netleague <- netleague(m.netmeta,
                           bracket = "[",
                           digits = 3,
                           details = TRUE,
                           seq = seq
                           )
    write.csv(netleague$random, paste0("./output/results/netleague/", df, ".csv"))
    
    # plots
    file <- paste0("./output/plots/forest/", df, ".svg")
    svg(file = file, width = 1200, height = 800)
    forest(m.netmeta)
    dev.off()

    assign(paste0(df), m.netmeta, envir = .GlobalEnv)

    dat <- data.frame(
        "k" = round(m.netmeta$k, 4),
        "TE" = round(m.netmeta$TE.random, 4),
        "ci.lb" = round(m.netmeta$lower.random, 4),
        "ci.ub" = round(m.netmeta$upper.random, 4),
        "pval" = round(m.netmeta$pval.random, 4)
    )
    
    write.csv(dat, paste0("./output/tables/network/", df, ".csv"))
    print(dat)    
#    assign(paste0(df, "_league"), netleague, envir = .GlobalEnv)
#    assign(paste0(df, "_rank"), netrank, envir = .GlobalEnv)
#    print(m.netmeta)
#    print(netleague)

}
```

**Network meta-analysis** for continuous outcomes

```r
continuous = c("ikdc_subjective", "lysholm", "tegner", "instrumented_laxity")
lapply(continuous, sm = "SMD", reference = "BPTB", network_meta_analysis)
```
