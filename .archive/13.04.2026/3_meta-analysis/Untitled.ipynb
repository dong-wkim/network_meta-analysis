```python
import numpy as np

def prop(n, x):
    n = n + 0.5
    x = x + 0.5
    p = x/n
    se = np.sqrt((p*(1-p))/n)
    return se
    
def sd(se, n):
    sd = se * np.sqrt(n)
    return sd
```

```python
print(sd(prop(86, 13), 86))
```

```python
prop(16, 4)
prop(20, 3)
prop(60, 0)
prop(60, 11)
prop(60, 2.2)
prop(60, 0)
prop(60, 10)
prop(60, 1.95)
```
