```python
import numpy as np

import numpy as np
import pandas as pd
import ipywidgets as widgets
from IPython.display import display

# assumes df already exists and has:
# id, first_author, second_author, num_authors, year, doi

df["id"] = df["id"].astype(int)
df = df.set_index("id", drop=False)

df["year_str"] = df["year"].astype("Int64").astype(str)

df["study_label"] = np.select(
    [
        df["num_authors"] == 1,
        df["num_authors"] == 2,
        df["num_authors"] > 2,
    ],
    [
        df["first_author"] + " (" + df["year_str"] + ")",
        df["first_author"] + " & " + df["second_author"] + " (" + df["year_str"] + ")",
        df["first_author"] + " et al. (" + df["year_str"] + ")",
    ],
    default=df["first_author"] + " (" + df["year_str"] + ")",
)

df["doi_url"] = np.where(
    df["doi"].notna() & (df["doi"].astype(str).str.strip() != ""),
    "https://doi.org/" + df["doi"].astype(str),
    "",
)

df["study_label_html"] = np.where(
    df["doi_url"] != "",
    '<a href="' + df["doi_url"] + '" target="_blank">'
    + df["study_label"]
    + "<sup>"
    + df["id"].astype(str)
    + "</sup></a>",
    df["study_label"] + "<sup>" + df["id"].astype(str) + "</sup>",
)

id_widget = widgets.IntSlider(
    value=int(df.index.min()),
    min=int(df.index.min()),
    max=int(df.index.max())
)

study_widget = widgets.HTML()

def update_study(change=None):
    row_id = id_widget.value
    study_widget.value = df.loc[row_id, "study_label_html"]

id_widget.observe(update_study, names="value")

update_study()

display(id_widget, study_widget)
```
