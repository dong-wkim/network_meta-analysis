Interactive data collection template for filling in meta-analysis data in a spreadsheet-like format.

```python
import pandas as pd
import ipywidgets as widgets
from ipydatagrid import DataGrid
from IPython.display import display
```

```python
template = pd.read_csv("./template.csv", encoding = "utf-8")
```

```python
columns = list(template.columns)
columns
```

```python
df = pd.DataFrame(columns = columns)
df.rename(columns = {"mi":"xi", "graft_type":"subgroup"}, inplace = True)
df["study"] = range(1, 100)
df = df.fillna("")
df.head()
```

```python
datagrid = DataGrid(
    df,
    editable = True,
    layout = {"height":"1000px", "width":"100%"})

display(datagrid)
```

```python
submit = widgets.Button(description = "Save", icon = "save")
display(submit)
```

```python

data = datagrid.data
data.to_csv("../meta-analysis/data/data.csv", encoding = "utf-8")
```
