```python

import ipywidgets as widgets
#from ipywidgets import Box, VBox, HBox
from IPython.display import display

box = widgets.Box()
box.layout.display = "flex"
box.layout.flex_flow = "center"
display(box)
```

```python
def hbox(words):
    items_layout = widgets.Layout(width = "100%")
    box_layout = widgets.Layout(display = "flex",
                                flex_flow = "row",
                                align_items = "stretch", 
                                align_content = "stretch",
                                width = "100%")
    
    items = [widgets.Button(description = str(w), layout = items_layout) for w in words]
    box = widgets.Box(children=items, layout=box_layout)
    display(box)

def vbox(words):
    items_layout = widgets.Layout(width = "90%", height = "40px")
    box_layout = widgets.Layout(display = "flex",
                                flex_flow = "column",
                                align_items = "stretch", 
                                align_content = "stretch",
                                width = "100px")
    
    items = [widgets.Button(description = str(w), layout = items_layout) for w in words]
    box = widgets.Box(children=items, layout=box_layout)
    display(box)

#vbox(words = ["Yes", "Maybe", "No"])
#(words = ["Yes", "Maybe", "No"])

def flexbox(words):
    items_layout = widgets.Layout(width = "90%", height = "40px")
    box_layout = widgets.Layout(display = "flex",
                                flex_flow = "center",
                                align_items = "stretch", 
                                align_content = "stretch",
                                width = "100px",
                                justify_content = "space_between")
    
    items = [widgets.Button(description = str(w), layout = items_layout) for w in words]
    box = widgets.Box(children=items, layout=box_layout)
    display(box)

a = hbox(words = ["Yes", "Maybe", "No"])
display(a)
```
