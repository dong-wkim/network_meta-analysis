```python
import ipywidgets as widgets
from ipyflex import FlexLayout
from IPython.display import display

def Slider(x = "float", y = None, value = 0, label = "", min = 0, max = 100, step = 0.1 ):
    if x == "int" and y == None:
        return widgets.IntSlider(
            value = float(value), 
            description = label)
    
    if x == "int" and y == "range":
        return widgets.IntRangeSlider(
            value = int(value),
            description = label)
    
    if x == "float" and y == None:
        return widgets.FloatSlider(
            value = float(value),
            description = label)
     
    if x == "float" and y == "log":
        return widgets.FloatLogSlider(
            value = float(value),
            description = label)
     
    if x == "float" and y == "range":  
        return widgets.FloatRangeSlider(
            value = float(value), 
            description = label)
```

```python
a = Slider(x = "float", y = None, value = 0, label = "", min = 0, max = 100, step = 0.1 )
display(a)
```

**progress(** x, value, label, min, max, step, color, orientation **)**

```python
def progress(x = "float", value = 0, label = "", min = 0, max = 100, step = 0.1, color = "green", orientation = "horizontal"):
    if x == "int":
        return widgets.IntProgress(
            value = int(value), description = label, min = min, max = max, step = step, style={'bar_color': f"{color}"}, orientation = orientation, )

    if x == "float":
        return widgets.FloatProgress(
            value = float(value), description = label, min = min, max = max, step = step, style={'bar_color': f"{color}"}, orientation = orientation)


# progress

g = progress("float", 75, "", 0, 100, 0.01, "green", orientation = "horizontal")
h = progress("int", 25, "", 0, 100, 0.01, "red", orientation = "horizontal")

display(g, h)
```

*i, j, k, l* = **textbox(** x, y, value, label **)**

```python
def Text(x = "bounded", y = "float", value = "", label = ""):
    if x == "bounded":
        if y == "float":
            return widgets.BoundedIntText(
        		value = float(value), 
                description = label
            )

        if y == "int":
            return widgets.BoundedFloatText(
        		value = float(value), 
                description = label
            )
        
    if x == None:
        if y == "float":
            return widgets.IntText(
            value = float(value), 
            description = label
        )
            
        if y == "int":
        	return widgets.FloatText(
        		value = float(value), 
                description = label
            )
```

```python
i = Text("bounded", "float", 0, "Label")
j = Text("bounded", "int", 0, "Label")
k = Text(None, "float", 0, "Label")
l = Text(None, "int", 0, "Label")

display(i, j, k, l)
```

```python
def Dropdown(options = "", label = ""):
	return widgets.Dropdown(
        options = options,
        value = options[0],
        description = label
    )

options = ["a", "b", "c"]
options = [
    "anterior cruciate ligament", 
    "bone-patellar tendon-bone", 
    "c"]

p = Dropdown(options, "")
display(p, widgets.Output())
```

```python
def RadioButtons(value = "", label = ""):
	return widgets.RadioButtons(
		value = float(value), 
        description = label
    )
```

```python
def Select(value = "", label = ""):
	return widgets.Select(
		value = float(value), 
        description = label
    )
```

```python
def SelectionSlider(value = "", label = ""):
	return widgets.SelectionSlider(
		value = float(value), 
        description = label
    )
```

```python
def SelectionRangeSlider(value = "", label = ""):
	return widgets.SelectionRangeSlider(
		value = float(value), 
        description = label
    )
```

```python
def ToggleButtons(options, value = "", label = "", icon = ["check"]):
	return widgets.ToggleButtons(
		options = options,
        value = value,
        description = label,
        icon = ['check']
    )

u = ToggleButtons(options = ["a", "b", "c", "d", "e"], value = "a", label = "")
display(u, widgets.Output())
```

```python
def SelectMultiple(options, value, label = ""):
	return widgets.SelectMultiple(
		options = options,
        value =(value,), 
        description = label)

v = SelectMultiple(options = ["a", "b", "c"], value = "a", label = "")
display(v, widgets.Output())
```

```python
def selectMultiple(value: 'Initial value', options: 'List', label: 'Label for valid'):
    return widgets.SelectMultiple(
        options=options,
        value=value,
        #rows=10,
        description=label,
        disabled=False
    )
    
m = selectMultiple(value = ["IKDC-SKF", "Pivot shift"], options = ["IKDC-SKF", "Lysholm", "Tegner", "Pivot shift", "Instrumental laxity", "Lachman", "Graft rupture"], label = "Outcome(s)")
display(m)
```
