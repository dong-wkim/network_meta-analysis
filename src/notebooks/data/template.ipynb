*i, j, k, l* = **textbox(** x, y, value, label **)**

```python
def Text(x = "bounded", y = "float", value = "", label = ""):
    if x == "bounded":
        if y == "float":
            return widgets.BoundedIntText(
        		value = float(value), 
                description = label,
                step = 0.1
            )

        if y == "int":
            return widgets.BoundedFloatText(
        		value = int(value), 
                description = label,
                step = 1
            )
        
    if x == None:
        if y == "float":
            return widgets.IntText(
            value = float(value), 
            description = label
        )
            
        if y == "int":
        	return widgets.FloatText(
        		value = float(value), 
                description = label
            )
```

```python
i = Text("bounded", "float", 0, "Label")
j = Text("bounded", "int", 0, "Label")
k = Text(None, "float", 0, "Label")
l = Text(None, "int", 0, "Label")
display(i, j, k, l, widgets.Output())
```

**boolean(** value, label **)**

```python
def ToggleButton(value = "True", label = 'Label'):
    return widgets.ToggleButton(
        value = value,
        description = label,
        button_style='', # 'success', 'info', 'warning', 'danger' or ''
        tooltip='Description',
        icon='check' # (FontAwesome names without the `fa-` prefix)
    )
```

```python
def Valid(value = "", label = ""):
	return widgets.Valid(
        value=value,
        description=label,
    )
```

```python
def Checkbox(value = False, label = ""):
    return widgets.Checkbox(
        value=value,
        description=label,
        disabled=False,
        indent=False
 )
```

```python
def Dropdown(options = "", label = str):
	return widgets.Dropdown(
        options = options,
        value = options[0],
        description = label
    )

data_types = ["continuous", "dichotomous", "ordinal", "time-to-event", "count"]

outcomes = ["IKDC-SKF", "Lysholm", "Tegner", 
            "Marx", "KOOS ADL", "KOOS QoL", "KOOS Sports", "KOOS Pain", "KOOS Symptoms",
            "Pivot shift", "Lachman", "Instrumental laxity",
            "Graft rupture", "arthrofibrosis", "anterior knee pain",
            "Extensor muscle strength recovery",
            "Flexor muscle strength recovery"]



subgroups = ["BPTB", "HT", "QT","PLT","AT","TA"]


p = Dropdown(subgroups, label = "subgroup(s)")
display(p)
outcomes = Dropdown(outcomes, "outcome(s)")

display(outcomes, widgets.Output())
display(widgets.FloatText(description = "sample size"))
display(widgets.FloatText(description = "mean"))
display(widgets.FloatText(description = "standard dev."))

```

```python
def RadioButtons(value = "", label = ""):
	return widgets.RadioButtons(
		value = float(value), 
        description = label
    )
```

```python
def Select(value = "", label = ""):
	return widgets.Select(
		value = float(value), 
        description = label
    )
```

```python
def SelectionSlider(value = "", label = ""):
	return widgets.SelectionSlider(
		value = float(value), 
        description = label
    )
```

```python
def SelectionRangeSlider(value = "", label = ""):
	return widgets.SelectionRangeSlider(
		value = float(value), 
        description = label
    )
```

```python
def ToggleButtons(options, value = "", label = "", icon = ["check"]):
	return widgets.ToggleButtons(
		options = options,
        value = value,
        description = label,
        icon = ['check']
    )

u = ToggleButtons(options = ["a", "b", "c", "d", "e"], value = "a", label = "", icon = 'check')
display(u, widgets.Output())
```

```python
def SelectMultiple(options, value, label = ""):
	return widgets.SelectMultiple(
		options = options,
        value =(value,), 
        description = label)

v = SelectMultiple(options = ["a", "b", "c"], value = "a", label = "")
display(v, widgets.Output())
```

```python
from ipywidgets import AppLayout, Button, Layout

def create_expanded_button(description, button_style):
    return Button(description=description, button_style=button_style, layout=Layout(height='100%', width='100%'))

header_button = create_expanded_button('Header', 'success')
center_button = create_expanded_button('Center', 'warning')
footer_button = create_expanded_button('Footer', 'success')

AppLayout(header=header_button,
          center=center_button,
          footer=footer_button)
```
